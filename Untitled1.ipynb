### 8. Test the ComfyUI API
*Example of how your frontend AI agent should send a request to ComfyUI.*

```python
import urllib.request
import json

# The API endpoint (Use your cloudflare URL if running from outside Colab)
URL = "http://127.0.0.1:8188/prompt"

# A basic ComfyUI workflow exported as API format
# This workflow uses the DreamShaper base model we downloaded in step 3
workflow_json = {
  "3": {
    "class_type": "KSampler",
    "inputs": {
      "seed": 123456789,
      "steps": 20,
      "cfg": 8,
      "sampler_name": "euler",
      "scheduler": "normal",
      "denoise": 1,
      "model": ["4", 0],
      "positive": ["6", 0],
      "negative": ["7", 0],
      "latent_image": ["5", 0]
    }
  },
  "4": {
    "class_type": "CheckpointLoaderSimple",
    "inputs": {
      "ckpt_name": "DreamShaper_8.safetensors"
    }
  },
  "5": {
    "class_type": "EmptyLatentImage",
    "inputs": {
      "width": 512,
      "height": 512,
      "batch_size": 1
    }
  },
  "6": {
    "class_type": "CLIPTextEncode",
    "inputs": {
      "text": "A beautiful 3D character, masterpiece, best quality",
      "clip": ["4", 1]
    }
  },
  "7": {
    "class_type": "CLIPTextEncode",
    "inputs": {
      "text": "bad anatomy, ugly, low resolution",
      "clip": ["4", 1]
    }
  },
  "8": {
    "class_type": "VAEDecode",
    "inputs": {
      "samples": ["3", 0],
      "vae": ["4", 2]
    }
  },
  "9": {
    "class_type": "SaveImage",
    "inputs": {
      "filename_prefix": "API_Test",
      "images": ["8", 0]
    }
  }
}

# Wrap the workflow in the "prompt" key
payload = {"prompt": workflow_json}
data = json.dumps(payload).encode('utf-8')

# Send the POST request
req = urllib.request.Request(URL, data=data, headers={'Content-Type': 'application/json'})

try:
    response = urllib.request.urlopen(req)
    result = json.loads(response.read().decode('utf-8'))
    print("✅ Success! Job queued on ComfyUI.")
    print("Response Data:", json.dumps(result, indent=2))
    print("\nYour frontend agent would now poll the server or wait for a websocket event to retrieve the generated image.")
except Exception as e:
    print("❌ Failed to reach API. Make sure ComfyUI is running.")
    print("Error:", e)
```

```python
import os
import subprocess
import urllib.request
import json
import time

# --- 1. Fast Custom Node Installation ---
NODES = "/content/ComfyUI/custom_nodes"
repos = [
    "https://github.com/ltdrdata/ComfyUI-Manager.git",
    "https://github.com/Kosinkadink/ComfyUI-VideoHelperSuite.git",
    "https://github.com/Lightricks/ComfyUI-LTXVideo.git",
    "https://github.com/kijai/ComfyUI-WanVideoWrapper.git",
    "https://github.com/kijai/ComfyUI-KJNodes.git",
    "https://github.com/kijai/ComfyUI_essentials.git",
    "https://github.com/Fannovel16/ComfyUI-Frame-Interpolation.git",
]

def run(command, cwd=None):
    subprocess.run(command, shell=True, check=True, cwd=cwd)

print("🚀 Installing high-speed nodes...")
for r in repos:
    name = r.rstrip("/").split("/")[-1].replace(".git", "")
    if not os.path.exists(f"{NODES}/{name}"):
        run(f"git clone {r}", cwd=NODES)

# Install requirements and sageattention
run("pip install -q sageattention")
print("✅ Nodes and speed backends ready.")

# --- 2. API Test Execution ---
print("\n📡 Testing API connection to background server...")
URL = "http://127.0.0.1:8188/prompt"

workflow_json = {
  "3": {"class_type": "KSampler", "inputs": {"seed": 123, "steps": 20, "cfg": 8, "sampler_name": "euler", "scheduler": "normal", "denoise": 1, "model": ["4", 0], "positive": ["6", 0], "negative": ["7", 0], "latent_image": ["5", 0]}},
  "4": {"class_type": "CheckpointLoaderSimple", "inputs": {"ckpt_name": "DreamShaper_8.safetensors"}},
  "5": {"class_type": "EmptyLatentImage", "inputs": {"width": 512, "height": 512, "batch_size": 1}},
  "6": {"class_type": "CLIPTextEncode", "inputs": {"text": "3D character render", "clip": ["4", 1]}},
  "7": {"class_type": "CLIPTextEncode", "inputs": {"text": "blurry, low quality", "clip": ["4", 1]}},
  "8": {"class_type": "VAEDecode", "inputs": {"samples": ["3", 0], "vae": ["4", 2]}},
  "9": {"class_type": "SaveImage", "inputs": {"filename_prefix": "Colab_Test", "images": ["8", 0]}}
}

payload = {"prompt": workflow_json}
data = json.dumps(payload).encode('utf-8')

try:
    req = urllib.request.Request(URL, data=data, headers={'Content-Type': 'application/json'})
    with urllib.request.urlopen(req, timeout=10) as response:
        result = json.loads(response.read().decode('utf-8'))
        print("🔥 SUCCESS! The API is working and the job is queued.")
        print(f"Prompt ID: {result['prompt_id']}")
except Exception as e:
    print(f"❌ Still getting a connection error: {e}")
    print("HINT: Ensure Cell #6 is currently running (it should finish in 1 second and leave the server in background).")
```
